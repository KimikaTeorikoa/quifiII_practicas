## Partikula kutxa batean
Liburutegia inportatu ondoren, sintaxi sinple batek hasierako balio baten eta amaierako balio baten arteko posizio kopuru bat sortzeko aukera ematen digu.
Lehenik eta behin, **partikula kutxa batean** landuko dugu, eta 0 eta $L=1$ artean definituko dugu.

```python
L = 1
steps = 2000
xvec = np.linspace(0, L, steps)
h = xvec[1] - xvec[0]
```

Erosotasunagatik, gure programan balio numeriko batzuk unitate gisa definituko ditugu, horrela ez baitugu horietaz arduratu beharko ondorengoetan.

```python
hbar = 1.0
m = 1.0
```

Osagai horiekin, Laplaziarraren bertsio matriziala osatzen has gaitezke.

```python
# create Laplacian via 3 point finite-difference method
Laplacian = (-2.0*np.diag(np.ones(steps)) + \
             np.diag(np.ones(steps-1),1) + \
             np.diag(np.ones(steps-1),-1))/(float)(h**2)
```

Kasu hain erraz honetan Hamiltondar osoa izateko, ez dugu Laplaziarrari $-\hbar^2/2m$ biderkatzea besterik behar.

```python
Hamiltonian = ((-0.5*(hbar**2)/m))*Laplacian
```

Orain, Hamiltondarraren bektore eta balio propioak lortzea besterik ez zaigu geratzen. Horretarako, `scipy` liburutegi berri bat erabiltzen dugu.
Erabiliz, nahi dugun eragiketa egin eta emaitzak azter ditzakegu.

```python
from scipy.linalg import eigh

def diagonalize_hamiltonian(Hamiltonian):
    return eigh(Hamiltonian)

evals, evecs = diagonalize_hamiltonian(Hamiltonian)
```

```python
print (" Valores más bajos para la energía:")
for i in range(6):
    print ('E(%g) = %.2f'%(i+1, evals[i]))
```
